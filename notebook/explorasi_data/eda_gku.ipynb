#### Cleaning Outlier

```python
import numpy as np

# 1. Define a sample year and month to inspect
sample_year = 2024.0
sample_month = 1

# 2. Filter the original df_combined_hourly DataFrame for this specific sample year and month
original_sample_group = df_combined_hourly[(df_combined_hourly['year'] == sample_year) &
                                             (df_combined_hourly['month'] == sample_month)]

print(f"--- Statistical Check for Year: {int(sample_year)}, Month: {sample_month} ---")

# 3. For each target column ('pm25', 'temperature', 'humidity'):
target_columns = ['pm25', 'temperature', 'humidity']

# Dictionary to store bounds for each column
bounds = {}

for col in target_columns:
    # a. Calculate Q1, Q3, and IQR for the original sample group's column data
    Q1 = original_sample_group[col].quantile(0.25)
    Q3 = original_sample_group[col].quantile(0.75)
    IQR = Q3 - Q1

    # b. Calculate the lower bound (Q1 - 1.5 * IQR) and upper bound (Q3 + 1.5 * IQR)
    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR

    bounds[col] = {'lower': lower_bound, 'upper': upper_bound}
    print(f"\nBounds for {col.capitalize()}:")
    print(f"  Q1: {Q1:.2f}, Q3: {Q3:.2f}, IQR: {IQR:.2f}")
    print(f"  Lower Bound: {lower_bound:.2f}")
    print(f"  Upper Bound: {upper_bound:.2f}")

# 4. Filter the df_combined_hourly_cleaned DataFrame for the same sample year and month
cleaned_sample_group = df_combined_hourly_cleaned[(df_combined_hourly_cleaned['year'] == sample_year) &
                                                    (df_combined_hourly_cleaned['month'] == sample_month)]

# 5. For each target column, check if any values in the cleaned sample group fall outside the calculated bounds
print("\n--- Outlier Check on Cleaned Data ---")
for col in target_columns:
    if not cleaned_sample_group[col].isnull().all():
        lower_bound = bounds[col]['lower']
        upper_bound = bounds[col]['upper']

        outliers_in_cleaned = cleaned_sample_group[(cleaned_sample_group[col] < lower_bound) |
                                                     (cleaned_sample_group[col] > upper_bound)]

        if not outliers_in_cleaned.empty:
            print(f"  WARNING: Outliers still present in cleaned {col} data for this sample!")
            print(outliers_in_cleaned[[col]])
        else:
            print(f"  SUCCESS: No outliers found in cleaned {col} data for this sample within the calculated 1.5 * IQR bounds.")
    else:
        print(f"  Skipping {col}: No non-null data in cleaned sample group for this column.")
```
